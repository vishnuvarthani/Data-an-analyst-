### Create a string, sent = ‘Coronavirus Caused Lockdowns Around The World.”

```python


sent = 'Coronavirus Caused Lockdowns Around The World.'
```

Use split function to convert the string into a list of words and save it in variable words and print the same

```python


words = sent.split()
print(words)
```

Convert each word in the list to lower case and store it in variable words_lower.
Print words_lower

```python


words_lower = [word.lower() for word in words]
print(words_lower)
```

Check whether ‘country’ is in the list

```python


print('country' in words_lower)
```

Remove the element ‘the’ from the list and print the list.

```python


words_lower.remove('the')
print(words_lower)
```

Select the first 4 words from the list words_lower using slicing and store them in a new variable x4

```python


x4 = words_lower[:4]
```

```python
# print x4

print(x4)
```

Convert the list of elements to single string using join function and print it

```python


print(' '.join(words_lower))
```

### Create colors = [‘red’,’blue’,’orange’], and fruits = [‘orange’,’grapes’,’apples’]

```python


colors = ['red', 'blue', 'orange']
fruits = ['orange', 'grapes', 'apples']
```

Print color and fruits

```python


print(colors, fruits)
```

Create colors_set, and fruits_set. (using set() ) and print them

```python


colors_set = set(colors)
fruits_set = set(fruits)
print(colors_set, fruits_set)
```

Find the union of both the sets.

```python


print(colors_set.union(fruits_set))
```

Find the intersection of both the sets

```python


print(colors_set.intersection(fruits_set))
```

Find the elements which are Fruits but not colors (using set.difference() )

```python


print(fruits_set.difference(colors_set))
```

```python


# The difference contains fruits that are not colors.
```

### Create temp = [17, ’Virat’, 50.0]

```python


temp = [17, 'Virat', 50.0]
```

Iterate through temp and print all the items in temp

```python


for item in temp:
    print(item)
```

replace first element with 11 in temp

```python


temp[0] = 11
```

Set temp1 = tuple(temp)

```python


temp1 = tuple(temp)
```

Iterate through temp1 and print all the items in temp1.

```python


for item in temp1:
    print(item)
```

replace first element with 17 in temp1

```python


# A tuple cannot be changed after creation.
try:
    temp1[0] = 17
except TypeError as error:
    print(error)
```

<b>Oops!! You got an error. Hey Don't worry! Its because Once a tuple is created, you cannot change its values unlike list.</b>

## DICTIONARY

### Create a dictionary d = {"actor":"amir","animal":"cat","earth":2,"list":[23,32,12]}

```python


d = {'actor': 'amir', 'animal': 'cat', 'earth': 2, 'list': [23, 32, 12]}
```

Print the value of d[0]

```python


# Missing key 0 raises KeyError, as described below.
try:
    print(d[0])
except KeyError as error:
    print('KeyError:', error)
```

<b>Oops!! again an error. again a fun fact. Dictionary return the value for key if key is in the dictionary, else throws KeyError
and we don't have key 0 here :( </b>

Store the value of d[‘actor’] to a new variable actor.
